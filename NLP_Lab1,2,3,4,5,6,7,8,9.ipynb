```python
#lab2(Word Generation)
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
import random


text = "The quick brown fox jumps over the lazy dog. The lazy dog, peeved to be left out, decides to jump over him."

# Tokenize the text
tokens = text.split()

# Create a dictionary of word pairs and their followers
word_pairs = {}
for i in range(len(tokens) - 2):
    pair = (tokens[i], tokens[i+1])
    follower = tokens[i+2]
    if pair not in word_pairs:
        word_pairs[pair] = []
    word_pairs[pair].append(follower)

# Choose a random starting pair
start_pair = random.choice(list(word_pairs.keys()))

# Generate a sentence
sentence = list(start_pair)
for i in range(10):
    next_word = random.choice(word_pairs[start_pair])
    sentence.append(next_word)
    start_pair = (start_pair[1], next_word)

# Convert the sentence back to a string
generated_text = ' '.join(sentence)

```

NLP LAB 5(NGRAM SMOOTHING)

```python
import nltk
from nltk import word_tokenize,ngrams
from nltk.lm import Laplace
from nltk.lm.preprocessing import padded_everygram_pipeline

#dload nltk data
nltk.download('punkt')
input_text = "The quick brown fox jumps over the lazy dog"

#Tokenize the words
tokens = word_tokenize(input_text)

#Define the value of n-grams
n=3

# Prepare data for the Laplace language model
train_data, padded_vocab = padded_everygram_pipeline(n, [tokens])

# Train a Laplace smoothing model
model = Laplace(n)
model.fit(train_data, padded_vocab)

# Define a new sentence and tokenize it
sentence = "The quick brown cat jumps over the lazy dog."
sentence_tokens = word_tokenize(sentence)

# Generate n-grams for the test sentence with padding
test_ngrams = list(ngrams(sentence_tokens, n, pad_left=True, pad_right=True, left_pad_symbol='<s>', right_pad_symbol='</s>'))

# Calculate the log probability for the n-grams in the sentence
log_probabilities = [model.logscore(ngram[-1], ngram[:-1]) for ngram in test_ngrams]

# Calculate the total log probability of the sentence (log-probability sum)
total_log_probability = sum(log_probabilities)

# Print the sentence and its log probability
print("Sentence:", sentence)
print("Log Probability:", total_log_probability)
```

NLP LAB6(POS Tagging) Hidden Markov Model

```python
import nltk
from nltk.corpus import brown
from nltk.tag import hmm
nltk.download('brown')
nltk.download('universal_tagset')
brown_corpus = brown.tagged_sents(tagset='universal')
split_ratio = 0.8
split_idx = int(len(brown_corpus) * split_ratio)
train_corpus = brown_corpus[:split_idx]
test_corpus = brown_corpus[split_idx:]
trainer = hmm.HiddenMarkovModelTrainer()
tagger = trainer.train_supervised(train_corpus)
accuracy = tagger.evaluate(test_corpus)
print("Accuracy:", accuracy)
```

NLP LAB7(Viterbi Decoding)

```python
import nltk
nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.corpus import brown

brown_tagged_sents = brown.tagged_sents(categories='news')
brown_sents = brown.sents(categories='news')
unigram_tagger = nltk.UnigramTagger(brown_tagged_sents)
bigram_tagger = nltk.BigramTagger(brown_tagged_sents, backoff=unigram_tagger)


def viterbi_decode(obs, states, start_p, trans_p, emit_p):
    V = [{}]
    for st in states:
        V[0][st] = {"prob": start_p[st] * emit_p[st].get(obs[0], 0), "prev": None}
    for t in range(1, len(obs)):
        V.append({})
        for st in states:
            max_tr_prob = max(V[t - 1][prev_st]["prob"] * trans_p[prev_st].get(st, 0) for prev_st in states)
            for prev_st in states:
                if V[t - 1][prev_st]["prob"] * trans_p[prev_st].get(st, 0) == max_tr_prob:
                    max_prob = max_tr_prob * emit_p[st].get(obs[t], 0)
                    V[t][st] = {"prob": max_prob, "prev": prev_st}
                    break
    opt = []
    max_prob = 0.0
    previous = None
    for st, data in V[-1].items():
        if data["prob"] > max_prob:
            max_prob = data["prob"]
            best_st = st
    opt.append(best_st)
    previous = best_st
    for t in range(len(V) - 2, -1, -1):
        opt.insert(0, V[t + 1][previous]["prev"])
        previous = V[t + 1][previous]["prev"]
    return opt


test_sentence = "The quick brown fox jumps over the lazy dog"
test_tokens = nltk.word_tokenize(test_sentence)
tagset = set([tag for sent in brown_tagged_sents for _, tag in sent])
test_tags = bigram_tagger.tag(test_tokens)
test_tags = [tag for _, tag in test_tags]
print(list(zip(test_tokens, test_tags)))
```

NLP LAB8(Building POS Tagger)

```python
from nltk import word_tokenize, pos_tag
from nltk.chunk import conlltags2tree, tree2conlltags
from nltk.tree import Tree

# Download required NLTK resources
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('conll2000')
#nltk.download('punkt_tab')

# Define a function to build the chunker
def build_chunker():
    # Define the grammar for the chunker
    grammar = "NP: {<DT>?<JJ>*<NN>}"

    # Create a chunk parser
    chunker = nltk.RegexpParser(grammar)

    return chunker

# Define a function to chunk a sentence
def chunk_sentence(sentence):
    # Tokenize the sentence
    tokens = word_tokenize(sentence)

    # Perform POS tagging
    tagged_tokens = pos_tag(tokens)

    # Create a chunker
    chunker = build_chunker()

    # Chunk the sentence
    chunked_tree = chunker.parse(tagged_tokens)

    return chunked_tree

# Test the chunker
sentence = "The quick brown fox jumps over the lazy dog."
chunked_tree = chunk_sentence(sentence)

# Print the chunked tree
print(chunked_tree)
```
